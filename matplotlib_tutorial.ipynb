# 📊 Matplotlib 绘图教程

欢迎来到 Matplotlib 数据可视化教程！本教程将从基础到进阶，带你学会用 Python 画各种漂亮的图表。

4. 设置图形样式与标签
5. 绘制多条曲线 + 图例
6. 保存图形到文件

```python
# 1️⃣ 导入必要的库
import matplotlib.pyplot as plt
import numpy as np

# 显示中文字体（如果系统有中文字体）
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

print("✅ 库导入成功！")
```

---
## 2️⃣ 创建示例数据

用 `numpy` 生成一些数据用于绘图。
- `np.linspace(start, stop, num)`：在指定区间生成等间隔的 num 个点
- 然后计算对应的 y 值

```python
# 2️⃣ 创建示例数据
x = np.linspace(0, 10, 100)      # 0 到 10 之间生成 100 个点
y1 = x ** 2                       # y = x²   (二次函数)
y2 = np.sin(x)                    # y = sin(x)  (正弦波)
y3 = np.cos(x)                    # y = cos(x)  (余弦波)

print(f"x 的范围: {x[0]:.1f} ~ {x[-1]:.1f}, 共 {len(x)} 个点")
print(f"y1 (x²) 范围: {y1[0]:.1f} ~ {y1[-1]:.1f}")
print(f"y2 (sin) 范围: {y2.min():.2f} ~ {y2.max():.2f}")
```

---
## 3️⃣ 绘制基本折线图

`plt.plot(x, y)` 是最基本的绘图函数。
- `x`：x 轴数据
- `y`：y 轴数据
- 第三个参数：格式字符串，如 `'r'`=红色, `'b'`=蓝色, `'g'`=绿色, `'--'`=虚线

```python
# 3️⃣ 绘制基本折线图

plt.figure(figsize=(8, 5))       # 设置画布大小 (宽, 高) 单位英寸
plt.plot(x, y1, 'b')             # 'b' = blue 蓝色实线
plt.title('基本折线图 - y = x²')   # 标题
plt.xlabel('x 轴')               # x 轴标签
plt.ylabel('y 轴')               # y 轴标签
plt.grid(True, alpha=0.3)       # 显示网格，透明度 0.3
plt.show()                       # 显示图形
```

---
## 4️⃣ 设置图形样式与标签

`plt.plot()` 的格式字符串可以组合使用：`[颜色][标记][线型]`

| 颜色 | 标记 | 线型 |
|------|------|------|
| `r` 红 | `o` 圆点 | `-` 实线 |
| `g` 绿 | `^` 三角 | `--` 虚线 |
| `b` 蓝 | `s` 方块 | `-.` 点划线 |
| `m` 紫红 | `*` 星号 | `:` 点线 |
| `k` 黑 | `x` 叉号 | |

```python
# 4️⃣ 各种样式演示

plt.figure(figsize=(12, 8))

# --- 子图 1：不同颜色和线型 ---
plt.subplot(2, 2, 1)  # 2行2列的第1个
plt.plot(x, y1, 'r-', label='y=x² (红实线)')     # 红色实线
plt.plot(x, y2, 'g--', label='sin(x) (绿虚线)')   # 绿色虚线
plt.plot(x, y3, 'b:', label='cos(x) (蓝点线)')    # 蓝色点线
plt.title('颜色 + 线型')
plt.legend()
plt.grid(True, alpha=0.3)

# --- 子图 2：带标记的曲线 ---
plt.subplot(2, 2, 2)
# 取少量点显示标记更清楚
x_sample = np.linspace(0, 10, 20)
plt.plot(x_sample, x_sample**2, 'ro-', label='圆点标记')    # 红色圆点实线
plt.plot(x_sample, np.sin(x_sample), 'gs--', label='方块标记') # 绿色方块虚线
plt.plot(x_sample, np.cos(x_sample), 'b^:', label='三角标记')  # 蓝色三角点线
plt.title('带标记的曲线')
plt.legend()
plt.grid(True, alpha=0.3)

# --- 子图 3：散点图 ---
plt.subplot(2, 2, 3)
np.random.seed(42)
scatter_x = np.random.randn(100)  # 100个随机数
scatter_y = np.random.randn(100)
colors = np.random.rand(100)
sizes = np.random.rand(100) * 200
plt.scatter(scatter_x, scatter_y, c=colors, s=sizes, alpha=0.6, cmap='viridis')
plt.title('散点图 (scatter)')
plt.colorbar(label='颜色映射')  # 颜色条
plt.grid(True, alpha=0.3)

# --- 子图 4：柱状图 ---
plt.subplot(2, 2, 4)
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 33]
colors_bar = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7']
plt.bar(categories, values, color=colors_bar, edgecolor='black', linewidth=1.5)
plt.title('柱状图 (bar)')
plt.ylabel('数值')

plt.suptitle('🎨 各种图形样式展示', fontsize=16, fontweight='bold')
plt.tight_layout()  # 自动调整子图间距
plt.show()
```

---
## 5️⃣ 绘制多条曲线 + 图例

用同一个 `plt.plot()` 画多条曲线，然后用 `plt.legend()` 添加图例。

💡 **提示**：`label` 参数设置每条线的名字，`legend()` 负责显示。

```python
# 5️⃣ 绘制多条曲线 + 图例

plt.figure(figsize=(10, 6))

# 画三条曲线，每条带不同的样式
plt.plot(x, y1, 'r-', linewidth=2, label='y = x²')           # 红色实线，粗2
plt.plot(x, y2, 'g--', linewidth=2, label='y = sin(x)')       # 绿色虚线，粗2
plt.plot(x, y3, 'b-.', linewidth=2, label='y = cos(x)')       # 蓝色点划线，粗2

# 添加一条参考线（y=0 水平线）
plt.axhline(y=0, color='gray', linestyle=':', alpha=0.5)

# 标注特殊点
plt.annotate('最高点', xy=(5, 25), xytext=(6, 30),
             arrowprops=dict(arrowstyle='->', color='red'),
             fontsize=12, color='red')

# 设置坐标轴范围
plt.xlim(0, 10)
plt.ylim(-2, 35)

# 标签和标题
plt.title('📈 多条曲线展示', fontsize=16, fontweight='bold')
plt.xlabel('X 轴', fontsize=12)
plt.ylabel('Y 轴', fontsize=12)

# 图例：loc 参数设置位置
plt.legend(loc='upper left', fontsize=11, framealpha=0.9)

plt.grid(True, alpha=0.3)
plt.show()
```

---
## 6️⃣ 保存图形到文件

用 `plt.savefig()` 可以把画好的图保存到文件。

常用格式：
- `PNG` — 通用图片格式
- `PDF` — 矢量格式，适合论文
- `SVG` — 网页矢量图
- `dpi` 参数控制分辨率

```python
# 6️⃣ 保存图形到文件

plt.figure(figsize=(10, 6))
plt.plot(x, y1, 'r-', linewidth=2, label='y = x²')
plt.plot(x, y2, 'g--', linewidth=2, label='y = sin(x)')
plt.plot(x, y3, 'b-.', linewidth=2, label='y = cos(x)')
plt.title('保存示例图形', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)

# 保存为 PNG（默认 dpi=100）
plt.savefig('my_plot.png', dpi=150, bbox_inches='tight')
print("✅ 已保存: my_plot.png (dpi=150)")

# 保存为 PDF（矢量格式）
plt.savefig('my_plot.pdf', bbox_inches='tight')
print("✅ 已保存: my_plot.pdf (矢量格式)")

plt.show()
```
